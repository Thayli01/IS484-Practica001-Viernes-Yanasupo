   ## Ejercicio 1: Agente de soporte

### Ficha PEAS

- **Percepción (S):** Tiempo de espera del ticket (min), nivel de urgencia reportado (baja, media, alta), si el cliente es premium.
- **Acciones (A):** Mantener en nivel 1, escalar a nivel 2, escalar a nivel 3
- **Entorno (E):** La mesa de ayuda de soporte técnico, con tickets entrando continuamente y distintos tipos de clientes
- **Objetivo:** Atender los casos más urgentes sin dejar esperando a ningún cliente
- **Medida de desempeño (P):** Tiempo promedio de resolución de tickets urgentes, satisfacción del cliente, cumplimiento de acuerdos de servicio (SLA) para clientes premium

## Justificación de las reglas

El factor más importante es la urgencia reportada por el cliente, una urgencia "alta " suele significar que el servicio está caido. El tiempo de espera se usa como corrector evitando que el sistema ignore casos que se van agravando. Ser cliente premium no debe pasar por encima de una urgencia baja real, pero sí acelera el escalamiento en casos de urgencia media, ya que esos clientes tienen acuerdos de servicio más estrictos.

### Código

```python
def agente_soporte(tiempo_espera_minutos, nivel_urgencia, cliente_premium):
    if nivel_urgencia == "alta":
        return "subir a nivel 3", "urgencia alta reportada por el cliente"
    elif nivel_urgencia == "media" and (tiempo_espera_minutos > 30 or cliente_premium):
        return "subir a nivel 2", "urgencia media con espera prolongada o cliente premium"
    elif nivel_urgencia == "media":
        return "mantener en nivel 1", "urgencia media dentro de tiempo aceptable"
    else:
        return "mantener en nivel 1", "urgencia baja"
```
